```python
'''Пример набора данных'''
data = [2, 4, 6, 8, 10, 12, 12, 14, 16, 18, 20]

'''Вычисление среднего'''
def calculate_mean(data):
    return sum(data) / len(data)

'''Вычисление медианы'''
def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2
    if n % 2 == 0:  # Четное количество элементов
        return (sorted_data[mid - 1] + sorted_data[mid]) / 2
    else:  # Нечетное количество элементов
        return sorted_data[mid]

'''Вычисление моды'''
def calculate_mode(data):
    frequency = {}
    for value in data:
        frequency[value] = frequency.get(value, 0) + 1
    max_freq = max(frequency.values())
    modes = [key for key, freq in frequency.items() if freq == max_freq]
    return modes

'''Вычисления'''
mean_value = calculate_mean(data)
median_value = calculate_median(data)
mode_value = calculate_mode(data)

'''Вывод результатов'''
print(f"Среднее: {mean_value}")
print(f"Медиана: {median_value}")
print(f"Мода: {mode_value}")
```

```python
import numpy as np
from scipy import stats

'''Пример данных для двух групп'''
group1 = [20, 22, 19, 24, 30, 27]
group2 = [30, 32, 29, 31, 28, 35]

'''Нулевая гипотеза (H0): Средние значения двух групп равны.'''
'''Альтернативная гипотеза (H1): Средние значения двух групп не равны.'''

'''Функция для выполнения t-теста'''
def perform_t_test(group1, group2):
    t_statistic, p_value = stats.ttest_ind(group1, group2)
    return t_statistic, p_value

'''Выполнение t-теста'''
t_statistic, p_value = perform_t_test(group1, group2)

'''Уровень значимости'''
alpha = 0.05

'''Вывод результатов'''
print(f"t-статистика: {t_statistic:.4f}")
print(f"p-значение: {p_value:.4f}")

'''Проверка гипотез'''
if p_value < alpha:
    print("Отвергаем нулевую гипотезу (H0). Есть статистически значимые различия.")
else:
    print("Не отвергаем нулевую гипотезу (H0). Нет статистически значимых различий.")
```

1. t-тест

```python
import numpy as np
from scipy import stats

'''Данные для двух групп'''
group1 = [75, 80, 85, 90]
group2 = [70, 75, 80, 85]

'''Функция для вычисления t-статистики и p-значения'''
def t_test(group1, group2):
    mean1 = np.mean(group1)
    mean2 = np.mean(group2)
    std1 = np.std(group1, ddof=1)
    std2 = np.std(group2, ddof=1)
    n1 = len(group1)
    n2 = len(group2)

    '''Объединенное стандартное отклонение'''
    sp = np.sqrt(((n1 - 1) * std1**2 + (n2 - 1) * std2**2) / (n1 + n2 - 2))

    '''t-статистика'''
    t_statistic = (mean1 - mean2) / (sp * np.sqrt(1/n1 + 1/n2))

    '''p-значение'''
    df = n1 + n2 - 2
    p_value = stats.t.sf(np.abs(t_statistic), df) * 2  # Двусторонний тест

    return t_statistic, p_value

'''Выполнение t-теста'''
t_statistic, p_value = t_test(group1, group2)

'''Вывод результатов'''
print("t-тест:")
print(f"t-статистика: {t_statistic:.4f}")
print(f"p-значение: {p_value:.4f}")
```

2. ANOVA (дисперсионный анализ)

```python
'''Данные для трех групп'''
group1 = [85, 90, 95]
group2 = [75, 80, 85]
group3 = [70, 75, 80]

'''Функция для вычисления ANOVA'''
def anova(groups):
    k = len(groups)  # Количество групп
    N = sum(len(group) for group in groups)  # Общее количество наблюдений

    '''Общая средняя'''
    overall_mean = np.mean([item for group in groups for item in group])

    '''Сумма квадратов между группами'''
    ssb = sum(len(group) * (np.mean(group) - overall_mean) ** 2 for group in groups)

    '''Сумма квадратов внутри групп'''
    ssw = sum(sum((x - np.mean(group)) ** 2 for x in group) for group in groups)

    '''Степени свободы'''
    dfb = k - 1
    dfw = N - k

    '''Средние квадратные значения'''
    msb = ssb / dfb
    msw = ssw / dfw

    '''F-статистика'''
    f_statistic = msb / msw

    return f_statistic, dfb, dfw

'''Выполнение ANOVA'''
f_statistic, dfb, dfw = anova([group1, group2, group3])

'''Вывод результатов'''
print("\nANOVA:")
print(f"F-статистика: {f_statistic:.4f}")
print(f"Степени свободы между группами: {dfb}")
print(f"Степени свободы внутри групп: {dfw}")
```

3. Критерий Манна-Уитни

Непараметрический тест для сравнения двух независимых выборок.

```python
import numpy as np
from scipy import stats

'''Генерация случайных данных для критерия Манна-Уитни'''
group1 = np.random.uniform(low=0, high=10, size=100)
group2 = np.random.uniform(low=5, high=15, size=100)

'''Критерий Манна-Уитни'''
u_stat, p_value = stats.mannwhitneyu(group1, group2)

print("\nКритерий Манна-Уитни:")
print(f"U-статистика: {u_stat:.3f}, p-значение: {p_value:.3f}")

if p_value < 0.05:
    print("Отвергаем нулевую гипотезу: группы различаются.")
else:
    print("Нет оснований отвергать нулевую гипотезу.")
```

12. Тест Тьюки (Tukey's HSD test)

- Множественное сравнение для определения, какие группы различаются друг от друга после ANOVA.


```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

'''Генерация данных для теста Тьюки'''
np.random.seed(42)
group1 = np.random.normal(loc=20, scale=5, size=100)
group2 = np.random.normal(loc=22, scale=5, size=100)
group3 = np.random.normal(loc=21, scale=5, size=100)

'''Объединение данных для теста Тьюки'''
data = np.concatenate([group1, group2, group3])
labels = ['Group 1'] * 100 + ['Group 2'] * 100 + ['Group 3'] * 100

'''Выполнение ANOVA'''
f_stat, p_value = stats.f_oneway(group1, group2, group3)
print("\nОднофакторная ANOVA (для Тьюки):")
print(f"F-статистика: {f_stat:.3f}, p-значение: {p_value:.3f}")

if p_value < 0.05:
    tukey_results = pairwise_tukeyhsd(data, labels)
    print("\nРезультаты теста Тьюки:")
    print(tukey_results)
else:
    print("Нет оснований для выполнения теста Тьюки, так как ANOVA не показала значительных различий.")
```

14. Анализ дисперсии (ANOVA)

- Ковариационный анализ (ANCOVA): Используется для сравнения средних значений с учетом влияния других переменных.
- Многомерный анализ дисперсии (MANOVA): Расширение ANOVA для множественных зависимых переменных.


```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

'''Генерация случайных данных'''
np.random.seed(42)
n = 100
group = np.random.choice(['A', 'B'], size=n)
covariate = np.random.normal(loc=5, scale=2, size=n)
outcome = 2 + 1.5 * (group == 'B') + 0.5 * covariate + np.random.normal(size=n)

'''Создание DataFrame'''
data = pd.DataFrame({'Group': group, 'Covariate': covariate, 'Outcome': outcome})

'''ANCOVA'''
model = ols('Outcome ~ C(Group) + Covariate', data=data).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

print("Результаты ANCOVA:")
print(anova_table)
```

```python
from statsmodels.multivariate.manova import MANOVA

'''Генерация случайных данных'''
np.random.seed(42)
n = 100
group = np.random.choice(['A', 'B'], size=n)
outcome1 = 2 + 1.5 * (group == 'B') + np.random.normal(size=n)
outcome2 = 1 + 1.0 * (group == 'B') + np.random.normal(size=n)

'''Создание DataFrame'''
data = pd.DataFrame({'Group': group, 'Outcome1': outcome1, 'Outcome2': outcome2})

'''MANOVA'''
manova = MANOVA.from_formula('Outcome1 + Outcome2 ~ C(Group)', data=data)
print("Результаты MANOVA:")
print(manova.mv_test())
```

- Модель структурных уравнений (SEM): Оценивает сложные взаимосвязи между переменными.


```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

'''Генерация случайных данных'''
np.random.seed(42)
n = 100
X1 = np.random.rand(n)
X2 = np.random.rand(n)
X3 = np.random.rand(n)
y = 3 + 2 * X1 + 1 * X2 - 1 * X3 + np.random.normal(size=n)

'''Создание DataFrame'''
data = pd.DataFrame({'X1': X1, 'X2': X2, 'X3': X3, 'y': y})

'''Множественная линейная регрессия'''
X = sm.add_constant(data[['X1', 'X2', 'X3']])  # Добавляем константу
model = sm.OLS(data['y'], X).fit()

print("Результаты множественной линейной регрессии:")
print(model.summary())
```

16. Тесты для категориальных данных

- Тест Фишера (Fisher's exact test): Используется для анализа зависимостей между двумя категориальными переменными в малых выборках.
- Критерий МакНемара (McNemar's test): Применяется для проверки изменений в бинарных данных.


```python
import numpy as np
from scipy import stats

'''Создание таблицы сопряженности (2x2)'''
'''Пример: 10 пациентов получили лечение и выздоровели, 5 не выздоровели'''
'''15 пациентов получили плацебо и 3 выздоровели, 12 не выздоровели'''
data = np.array([[10, 5],
                 [3, 12]])

'''Тест Фишера'''
odds_ratio, p_value = stats.fisher_exact(data)

print("Результаты теста Фишера:")
print(f"Odds Ratio: {odds_ratio:.3f}, p-значение: {p_value:.3f}")

if p_value < 0.05:
    print("Отвергаем нулевую гипотезу: есть зависимость между переменными.")
else:
    print("Нет оснований отвергать нулевую гипотезу.")
```

```python
from statsmodels.stats.contingency_tables import mcnemar

'''Создание таблицы сопряженности (2x2)'''
'''Пример: до и после лечения'''
'''20 пациентов до лечения были больны (1) и 5 выздоровели (0)'''
'''5 пациентов после лечения были больны (1) и 10 выздоровели (0)'''
data = np.array([[20, 5],
                 [5, 10]])

'''Критерий МакНемара'''
result = mcnemar(data, exact=True)

print("\nРезультаты критерия МакНемара:")
print(f"Statistic: {result.statistic:.3f}, p-значение: {result.pvalue:.3f}")

if result.pvalue < 0.05:
    print("Отвергаем нулевую гипотезу: есть изменения между двумя временными точками.")
else:
    print("Нет оснований отвергать нулевую гипотезу.")
```

```python
import numpy as np
import pandas as pd
from statsmodels.tsa.stattools import adfuller

'''Генерация случайного временного ряда'''
np.random.seed(42)
data = np.random.normal(size=100).cumsum()  # Нестационарный ряд

'''Применение теста Дики-Фуллера'''
adf_result = adfuller(data)

print("Результаты теста Дики-Фуллера:")
print(f"ADF Statistic: {adf_result[0]:.3f}")
print(f"p-значение: {adf_result[1]:.3f}")
print(f"Количество лагов: {adf_result[2]}")
print(f"Число наблюдений: {adf_result[3]}")
print("Критические значения:")
for key, value in adf_result[4].items():
    print(f"  {key}: {value:.3f}")

if adf_result[1] < 0.05:
    print("Отвергаем нулевую гипотезу: ряд стационарный.")
else:
    print("Нет оснований отвергать нулевую гипотезу: ряд нестационарный.")
```

- Нормальное распределение (распределение Гаусса) характеризуется симметричной колоколообразной формой графика плотности вероятности. Это один из важнейших типов распределений в статистике.
- Экспоненциальное распределение используется для моделирования случайных величин, связанных с интервалами между редкими событиями.

Ковариация и корреляция

Ковариация измеряет степень совместного изменения двух переменных. Положительная ковариация означает, что обе переменные имеют тенденцию увеличиваться вместе, отрицательная — одна увеличивается, когда другая уменьшается. Однако абсолютная величина ковариации зависит от масштаба самих переменных, что делает её интерпретацию сложной.

Корреляция нормализует ковариацию путём деления на произведение стандартных отклонений обеих переменных. Коэффициент корреляции Пирсона принимает значения от −1−1 до 11, где:

Расчёт медианы (median)

```python
def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)

    if n % 2 == 1:
        # Нечётное количество элементов
        median = sorted_data[n // 2]
    else:
        # Четное количество элементов
        mid1 = sorted_data[(n // 2) - 1]
        mid2 = sorted_data[n // 2]
        median = (mid1 + mid2) / 2

    return median

median_value = calculate_median(data)
print(f"Медиана: {median_value}")
```

Расчёт моды (mode)

```python
from collections import Counter

def calculate_mode(data):
    counts = Counter(data)
    max_count = max(counts.values())
    modes = [num for num, count in counts.items() if count == max_count]
    return modes

modes = calculate_mode(data)
if len(modes) > 1:
    print(f"Набор бимодален/мультимодален: {modes}")
elif len(modes) == 1:
    print(f"Мода: {modes[0]}")
else:
    print("Нет моды")
```

Расчёт дисперсии (variance)

```python
def calculate_variance(data):
    mean_val = calculate_mean(data)
    variance = sum((xi - mean_val) ** 2 for xi in data) / (len(data) - 1)
    return variance

variance_value = calculate_variance(data)
print(f"Дисперсия: {variance_value:.2f}")
```

Оценка ковариации и коэффициента корреляции

```python
# Новые наборы данных
X = [1, 2, 3, 4]
Y = [2, 4, 6, 8]

# Функция расчета ковариации
def covariance(X, Y):
    x_mean = calculate_mean(X)
    y_mean = calculate_mean(Y)
    covar = sum((xi - x_mean)*(yi - y_mean) for xi, yi in zip(X, Y)) / (len(X)-1)
    return covar

cov_xy = covariance(X, Y)
print(f"Ковариация X и Y: {cov_xy:.2f}")

# Стандартное отклонение
std_dev_x = np.std(X, ddof=1)
std_dev_y = np.std(Y, ddof=1)

# Коэффициент корреляции Пирсона
correlation_coefficient = cov_xy / (std_dev_x * std_dev_y)
print(f"Коэффициент корреляции Пирсона: {correlation_coefficient:.2f}")
```

1. Тест на нормальность:

- Тест Андресона-Дарлинга (Anderson-Darling test)
- Тест Лиллиефорса (Lilliefors test)

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# Генерация двух выборок: одна нормальная, другая равномерная
np.random.seed(42)
mu, sigma = 5, np.sqrt(2)
sample_size = 1000
normal_data = np.random.normal(mu, sigma, sample_size)
uniform_data = np.random.uniform(-1, 1, size=sample_size)

# Проверка нормальной выборки с использованием Anderson-Darling test
result_normal = stats.anderson(normal_data, dist='norm')

# Проверка равномерной выборки с использованием Kolmogorov-Smirnov test
ks_result_uniform = stats.kstest(uniform_data, 'uniform', args=(-1, 2))  # uniform(a,b), a=-1, b=1

# Интерпретация результатов
def interpret_anderson_results(results):
    stat, crit_vals, sig_levels = results
    print(f'Статистика A²: {stat:.4f}')
    for i in range(len(sig_levels)):
        alpha = sig_levels[i]
        crit_val = crit_vals[i]
        decision = 'гипотеза отвергается' if stat > crit_val else 'гипотеза принимается'
        print(f'{decision} на уровне значимости α={alpha:.2f}')

def interpret_ks_test(pvalue, alpha=0.05):
    if pvalue >= alpha:
        return f'Гипотеза о равномерности принимается (p-значение: {pvalue:.4f})'
    else:
        return f'Гипотеза о равномерности отвергается (p-значение: {pvalue:.4f})'

# Анализ результатов тестов
print("Проверка нормального распределения с помощью Anderson-Darling:\n")
interpret_anderson_results(result_normal)

print("\nПроверка равномерного распределения с помощью Kolmogorov-Smirnov:\n")
print(interpret_ks_test(ks_result_uniform.pvalue))

# Гистограмма для визуального представления данных
fig, axs = plt.subplots(1, 2, figsize=(12, 5))
axs[0].hist(normal_data, bins=30, density=True, color="skyblue")
axs[0].set_title('Нормальное распределение')
axs[0].set_xlabel('Значения')
axs[0].set_ylabel('Плотность')

axs[1].hist(uniform_data, bins=30, density=True, color="lightgreen")
axs[1].set_title('Равномерное распределение')
axs[1].set_xlabel('Значения')
axs[1].set_ylabel('Плотность')

plt.tight_layout()
plt.show()
```

```python
# Тест Лиллиефорса
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statsmodels.stats.diagnostic import lilliefors

# Задание семени генератора случайных чисел для повторяемости
np.random.seed(42)

# Создание синтетических наборов данных с разными степенями отклонений от нормы
datasets = {
    'нормально': np.random.normal(loc=0, scale=1, size=1000),
    'экспоненциально': np.random.exponential(scale=1, size=1000),
    'равномерно': np.random.uniform(low=-1, high=1, size=1000),
    'лог-нормально': np.random.lognormal(mean=0, sigma=1, size=1000),
}

# Функция вывода результатов теста Лиллиефорса
def run_lilliefors_test(dataset_name, data):
    statistic, p_value = lilliefors(data)
    df = pd.DataFrame({'values': data})
    sns.distplot(df['values'], kde=True, label=f"{dataset_name}, p-value: {p_value:.4f}")
    plt.legend()
    plt.title(f"Плотность распределения для {dataset_name}\n Статистика Lilliefors: {statistic:.4f}")
    plt.show()

    # Интерпретация результата
    if p_value <= 0.05:
        print(f"Гипотеза о нормальности распределения для '{dataset_name}' отвергнута.")
    else:
        print(f"Нет оснований отвергнуть гипотезу о нормальности распределения для '{dataset_name}'.")

# Запускаем тесты на всех наборах данных
for dataset_name, data in datasets.items():
    print(f"\n--- Тест Лиллиефорса для набора данных '{dataset_name}' ---")
    run_lilliefors_test(dataset_name, data)
```

2. Тесты для проверки равенства дисперсий:

- Тест Флевена (Levene's test)
- Тест Бартлетта (Bartlett's test)



```python
# Тест Флевена
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import levene
import seaborn as sns

# Генерируем синтетические данные
np.random.seed(42)

# Данные первой группы с низкой дисперсией
group1 = np.random.normal(loc=10, scale=1, size=100)

# Вторая группа с большей дисперсией
group2 = np.random.normal(loc=10, scale=2, size=100)

# Третья группа также с большой дисперсией
group3 = np.random.normal(loc=10, scale=3, size=100)

# Объединяем группы в список
groups = [group1, group2, group3]

# Определяем центральный момент (mean - среднее значение, median - медиану)
central_moment = 'median'

# Выполняем тест Левене
statistic, p_value = levene(*groups, center=central_moment)

# График распределения данных
sns.set_style("whitegrid")
plt.figure(figsize=(10, 6))
for idx, group in enumerate(groups):
    sns.kdeplot(group, shade=True, label=f'Группа {idx+1}', linewidth=2)
plt.title('Плотности распределения данных')
plt.xlabel('Значения')
plt.ylabel('Плотность')
plt.legend()
plt.show()

# Интерпретация результатов
if p_value < 0.05:
    print(f"Тест Левене показал наличие значительных различий в дисперсиях (F-статистика: {statistic:.4f}, p-значение: {p_value:.4f}). Гипотеза равенства дисперсий отвергнута.")
else:
    print(f"Тест Левене не выявил значимых различий в дисперсиях (F-статистика: {statistic:.4f}, p-значение: {p_value:.4f}). Нет оснований отвергнуть гипотезу равенства дисперсий.")
```

```python
# Тест Бартлетта
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bartlett


def generate_groups(num_groups=3, mean=0, std_devs=[1, 2, 3], n_samples=100):
    """
    Генерирует несколько групп данных с указанными стандартными отклонениями.
    :param num_groups: Количество групп
    :param mean: Среднее значение
    :param std_devs: Список стандартных отклонений для каждой группы
    :param n_samples: Размер выборки для каждой группы
    :return: Списки данных
    """
    groups = []
    for sd in std_devs[:num_groups]:
        group = np.random.normal(loc=mean, scale=sd, size=n_samples)
        groups.append(group)
    return groups


def plot_group_distributions(groups):
    """
    Рисует гистограммы для каждой группы данных.
    """
    plt.figure(figsize=(10, 6))
    colors = ['b', 'g', 'r']
    labels = [f'Группа {i+1}' for i in range(len(groups))]
    for i, group in enumerate(groups):
        plt.hist(group, bins=30, alpha=0.5, color=colors[i % len(colors)], label=labels[i])
    plt.legend()
    plt.title('Группы данных')
    plt.xlabel('Значения')
    plt.ylabel('Количество наблюдений')
    plt.grid(True)
    plt.show()


def perform_bartlett_test(groups):
    """
    Проводит Bartlett-тест и выводит результаты.
    """
    statistic, p_value = bartlett(*groups)
    print(f'Statistic: {statistic:.4f}, P-value: {p_value:.4f}')
    if p_value < 0.05:
        print("Дисперсии существенно отличаются!")
    else:
        print("Нет оснований утверждать, что дисперсии существенно отличаются.")


# Параметры генерации данных
num_groups = 3      # Число групп
std_deviations = [1, 2, 3]  # Стандартные отклонения для каждой группы
samples_per_group = 100       # Размер выборки для каждой группы

# Генерация данных
groups = generate_groups(num_groups=num_groups, std_devs=std_deviations, n_samples=samples_per_group)

# Вывод графиков распределения
plot_group_distributions(groups)

# Проведение теста Бартлетта
perform_bartlett_test(groups)
```

3. Непараметрические тесты:
   - Тест для независимых выборок (Mann-Whitney U test)
   - Тест Вилкоксона для парных выборок (Wilcoxon signed-rank test)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import mannwhitneyu, wilcoxon

# Генерация большого объема данных
np.random.seed(42)

# Группа 1: Номинальная средняя величина с небольшим разбросом
group1 = np.random.normal(loc=10, scale=1, size=1000)

# Группа 2: Средняя чуть меньше, немного больший разброс
group2 = np.random.normal(loc=9.5, scale=1.5, size=1000)

# Группа 3: Значительно отличается от первых двух
group3 = np.random.normal(loc=12, scale=2, size=1000)

# Тест Манна-Уитни U (для независимых выборок)
def compare_with_mann_whitney(g1, g2):
    u_stat, p_value = mannwhitneyu(g1, g2, alternative='two-sided')
    print(f"Mann-Whitney U-test: Statistic = {u_stat:.4f}, p-value = {p_value:.4f}")
    if p_value < 0.05:
        print("Есть значительные различия между группами.\n")
    else:
        print("Невозможно сказать, что различия существенны.\n")

# Тест Вилкоксона (для зависимых выборок)
def compare_with_wilcoxon(d1, d2):
    w_stat, p_value = wilcoxon(d1, d2)
    print(f"Wilcoxon signed-rank test: Statistic = {w_stat:.4f}, p-value = {p_value:.4f}")
    if p_value < 0.05:
        print("Есть значительные различия между двумя условиями.\n")
    else:
        print("Невозможно сказать, что различия существенны.\n")

# Проверка попарно всех трех групп
compare_with_mann_whitney(group1, group2)
compare_with_mann_whitney(group1, group3)
compare_with_mann_whitney(group2, group3)

# Тест Вилкоксона (создаём искусственно связанные пары данных)
diff_data1 = group1 + np.random.randn(len(group1)) / 10  # Небольшое изменение первого набора
diff_data2 = group2 + np.random.randn(len(group2)) / 10  # Небольшое изменение второго набора

compare_with_wilcoxon(diff_data1, diff_data2)

# Визуализация данных
plt.figure(figsize=(10, 6))
plt.subplot(1, 2, 1)
plt.boxplot([group1, group2, group3], vert=False)
plt.title('Boxplots of Data Groups')
plt.xlabel('Values')

plt.subplot(1, 2, 2)
plt.hist([group1, group2, group3], histtype='stepfilled', alpha=0.5, label=['Group 1', 'Group 2', 'Group 3'])
plt.legend()
plt.title('Histogram Comparison')
plt.xlabel('Values')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

4. Корреляционный анализ:
   - Коэффициент Пирсона (Pearson correlation coefficient)
   - Коэффициент Спирмена (Spearman's rank correlation coefficient)

```python
# Коэффициент Пирсона
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# Генерация синтетических данных с известной корреляционной структурой
np.random.seed(42)
N = 1000
x = np.linspace(-10, 10, N)
y = 2*x + np.random.normal(size=N)*3  # Добавили шум для реалистичного эффекта
z = -0.5*y + np.random.normal(size=N)*2  # Ещё одно зависимое измерение

df = pd.DataFrame({
    'X': x,
    'Y': y,
    'Z': z
})

# Расчёт матрицы корреляций
correlation_matrix = df.corr(method='pearson')

# Тепловая карта корреляций
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Матрица корреляций Пирсона')
plt.show()

# Точечная диаграмма для конкретной пары X-Y
plt.scatter(x, y, s=20, c='b', marker='o', edgecolors='none', alpha=0.7)
plt.title('Корреляционная связь между X и Y')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.show()

# Коэффициент корреляции и p-значение для одной пары переменных
corr_xy, p_value_xy = pearsonr(x, y)
print(f"Корреляция Пирсона между X и Y: {corr_xy:.4f}, p-значение: {p_value_xy:.4f}")

# Фильтрация сильных корреляционных связей
strong_correlations = correlation_matrix[(abs(correlation_matrix) > 0.5)]
print("\nСильные корреляционные связи (|корреляция| > 0.5):\n", strong_correlations)
```

```python
# Коэффициент Спирмена
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import spearmanr

# Генерация синтетических данных
np.random.seed(42)
N = 1000
x = np.linspace(-10, 10, N)
y_linear = 2*x + np.random.normal(size=N)*3  # Линейная зависимость с шумом
y_quadratic = x**2 + np.random.normal(size=N)*10  # Квадратичная зависимость с шумом
y_random = np.random.normal(size=N)*5  # Случайные данные

# Формирование датасета
df = pd.DataFrame({
    'X': x,
    'Linear_Y': y_linear,
    'Quadratic_Y': y_quadratic,
    'Random_Y': y_random
})

# Вычислим матрицу ранговой корреляции Спирмена
correlation_matrix = df.corr(method='spearman')

# Постройте тепловую карту
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Матрица ранговой корреляции Спирмена')
plt.show()

# Определение силы корреляции
def strength_of_corr(corr_value):
    abs_corr = abs(corr_value)
    if abs_corr < 0.3:
        return 'слабый'
    elif abs_corr < 0.7:
        return 'умеренно сильный'
    else:
        return 'очень сильный'

# Рассчитываем корреляцию Спирмена для конкретной пары признаков
corr_xy, p_value_xy = spearmanr(x, y_linear)
print(f"Сила корреляции между X и Linear_Y: {strength_of_corr(corr_xy)} (коэффициент: {corr_xy:.4f}), p-значение: {p_value_xy:.4f}")

# Проверка влияния шума на корреляцию
noisy_y = y_linear + np.random.normal(size=N)*10  # Добавляем дополнительный шум
corr_noisy, _ = spearmanr(x, noisy_y)
print(f"Корреляция после добавления сильного шума: {corr_noisy:.4f}")

# Понимание важности наличия линейности
corr_quadratic, _ = spearmanr(x, y_quadratic)
print(f"Корреляция между X и Quadratic_Y (квадратичная зависимость): {corr_quadratic:.4f}")
```

5. Регрессионные модели:
   - Логистическая регрессия (Logistic regression)
   - Полиномиальная регрессия (Polynomial regression)

```python
# Логистическая регрессия
import numpy as np
import pandas as pd
import statsmodels.api as sm
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, roc_auc_score
from sklearn.linear_model import LogisticRegression

# 1. Генерация синтетических данных
np.random.seed(42)
N = 1000
X1 = np.random.normal(0, 1, N)
X2 = np.random.normal(0, 1, N)
X3 = np.random.binomial(n=1, p=0.5, size=N)  # Категоричская переменная

# Генерируем целевую переменную с некоторым влиянием факторов
y = (X1 + X2 + X3*2 + np.random.normal(0, 1, N)) > 0

# Собираем данные в dataframe
data = pd.DataFrame({'X1': X1, 'X2': X2, 'X3': X3, 'Target': y.astype(int)})

# Преобразование категориальных переменных
data = pd.get_dummies(data, drop_first=True)

# Разделение на тренировочную и тестовую части
train_data, test_data = train_test_split(data, test_size=0.2, random_state=42)

# Отделяем признаки и цель
X_train = train_data.drop(columns=['Target'])
y_train = train_data['Target']
X_test = test_data.drop(columns=['Target'])
y_test = test_data['Target']

# Масштабирование признаков
scaler = StandardScaler().fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Добавляем константу для смещения (intercept)
X_train_sm = sm.add_constant(X_train_scaled)
X_test_sm = sm.add_constant(X_test_scaled)

# Обучение базовой логистической регрессии
logit_model = sm.Logit(y_train, X_train_sm)
result = logit_model.fit()
print(result.summary())

# Прогноз на тестовом наборе
test_pred_probs = result.predict(X_test_sm)
test_pred_class = (test_pred_probs > 0.5).astype(int)

# Оценка точности модели
print(classification_report(y_test, test_pred_class))
roc_auc = roc_auc_score(y_test, test_pred_probs)
print(f"AUC ROC Score: {roc_auc:.4f}")

# Регуляризация с помощью scikit-learn
clf = LogisticRegression(C=1.0, penalty='l2')  # C - обратная сила регуляризации
clf.fit(X_train_scaled, y_train)
pred_prob_sklearn = clf.predict_proba(X_test_scaled)[:, 1]
print(f"Roc-AUC (sklearn): {roc_auc_score(y_test, pred_prob_sklearn):.4f}")

# Оптимизация гиперпараметров с помощью grid search
parameters = {'C': [0.01, 0.1, 1, 10]}
gs = GridSearchCV(LogisticRegression(), parameters, scoring='roc_auc', cv=5)
gs.fit(X_train_scaled, y_train)
best_clf = gs.best_estimator_

# Прогноз и оценка итогового результата
final_pred_prob = best_clf.predict_proba(X_test_scaled)[:, 1]
final_auc = roc_auc_score(y_test, final_pred_prob)
print(f"Optimized Roc-AUC: {final_auc:.4f}")
```

```python
# Полиномиальная регрессия
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import cross_validate, KFold

# 1. Генерация синтетических данных с нелинейной зависимостью и шумом
np.random.seed(42)
X = np.sort(np.random.rand(50) * 10)  # Независимая переменная
true_function = lambda x: np.sin(x) + 0.5*np.cos(x)**2  # Истинная функция
y = true_function(X) + np.random.normal(0, 0.2, size=X.shape)  # Целевые значения с шумом

# 2. Подбор оптимальной степени полинома с перекрестной проверкой
degrees = list(range(1, 10))  # Степени полинома для перебора
kf = KFold(n_splits=5, shuffle=True, random_state=42)

scores = {}
for degree in degrees:
    pipeline = make_pipeline(PolynomialFeatures(degree), Ridge(alpha=1.0))
    cv_scores = cross_validate(pipeline, X.reshape(-1, 1), y, cv=kf, scoring='neg_mean_squared_error')
    scores[degree] = -cv_scores['test_score'].mean()  # Сохраняем отрицательные среднеквадратичные ошибки

optimal_degree = min(scores, key=lambda x: scores[x])  # Лучшая степень полинома
print(f"Лучшая степень полинома: {optimal_degree}")

# 3. Применение оптимального полинома и обучение модели
pipeline = make_pipeline(PolynomialFeatures(optimal_degree), Ridge(alpha=1.0))
pipeline.fit(X.reshape(-1, 1), y)

# 4. Визуализация результатов
X_plot = np.linspace(min(X)-1, max(X)+1, 100)  # Широкий диапазон для экстраполирования
y_predicted = pipeline.predict(X_plot.reshape(-1, 1))

plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='darkorange', label='Исходные точки')
plt.plot(X_plot, y_predicted, color='navy', lw=2, label=f'Аппроксимация ({optimal_degree}-степень)')
plt.plot(X_plot, true_function(X_plot), '--', color='gray', label='Истинная функция')
plt.legend()
plt.title('Полиномиальная регрессия с регуляризацией Ridge')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()

# 5. Предсказание новой точки вне диапазона
new_point = 12.0
predicted_new = pipeline.predict([[new_point]])
print(f"Предсказанное значение для X={new_point}: {predicted_new[0]:.4f}")
```

6. Модели времени:
   - ARIMA (Autoregressive Integrated Moving Average)
   - Модели сезонности и тренда

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.statespace.sarimax import SARIMAX
from itertools import product

# 1. Генерация синтетических данных с сезонностью
np.random.seed(42)
N = 100
trend = np.arange(N)
seasonality = np.sin(2 * np.pi * trend / 12)  # Годовая сезонность
noise = np.random.normal(0, 0.5, N)
sales = trend + seasonality + noise

# Дата-индексация
dates = pd.date_range(start='2020-01-01', periods=N, freq='MS')
series = pd.Series(sales, index=dates)

# 2. Автоопределение лучших параметров (p, d, q)
ps = range(0, 3)
ds = range(0, 2)
qs = range(0, 3)
pdq = list(product(ps, ds, qs))

best_aic = float("inf")
best_order = None
best_seasonal_order = None

for param in pdq:
    for seasonal_param in [(1, 0, 1, 12)]:  # Попробуем одну конфигурацию сезонности
        try:
            model = SARIMAX(series, order=param, seasonal_order=seasonal_param)
            results = model.fit(disp=False)
            current_aic = results.aic
            if current_aic < best_aic:
                best_aic = current_aic
                best_order = param
                best_seasonal_order = seasonal_param
        except Exception as e:
            continue

print(f"Лучшие параметры: порядок={best_order}, сезонный порядок={best_seasonal_order}, AIC: {best_aic}")

# 3. Обучение финальной модели
final_model = SARIMAX(series, order=best_order, seasonal_order=best_seasonal_order)
results = final_model.fit(disp=False)

# 4. Диагностика модели
residuals = results.resid
fig, axes = plt.subplots(2, 1, figsize=(10, 8))
axes[0].plot(residuals)
axes[0].set_title('Остатки модели')
plot_acf(residuals, ax=axes[1])
plt.show()

# 5. Прогнозирование
forecast_steps = 12  # Прогноз на 12 месяцев вперёд
forecast = results.forecast(steps=forecast_steps)
confidence_intervals = results.get_forecast(steps=forecast_steps).conf_int()

# 6. Визуализация прогноза
plt.figure(figsize=(10, 6))
plt.plot(series, label='Фактические данные', color='black')
plt.plot(series.index[-len(confidence_intervals):], confidence_intervals.iloc[:, 0], label='Нижняя граница доверительного интервала', color='grey', ls='--')
plt.plot(series.index[-len(confidence_intervals):], confidence_intervals.iloc[:, 1], label='Верхняя граница доверительного интервала', color='grey', ls='--')
plt.plot(pd.date_range(start=series.index[-1]+pd.DateOffset(months=1), periods=forecast_steps, freq='MS'), forecast, label='Прогноз', color='red')
plt.legend()
plt.title('SARIMA Прогноз временных рядов')
plt.show()
```

7. Кластерный анализ:
   - K-средние (K-means clustering)
   - Иерархическая кластеризация (Hierarchical clustering)

```python
# K-средние
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, calinski_harabasz_score

# 1. Генерация синтетических данных
np.random.seed(42)
centers = [[1, 1], [-1, -1], [1, -1]]
cluster_std = [0.3, 0.5, 0.4]
X, y_true = make_blobs(n_samples=300, centers=centers, cluster_std=cluster_std)

# 2. Подбор оптимального числа кластеров (Elbow Method & Silhouette Analysis)
inertias = []
silhouettes = []
range_n_clusters = range(2, 10)

for n_clusters in range_n_clusters:
    kmeans = KMeans(n_clusters=n_clusters, init='k-means++', random_state=42)
    kmeans.fit(X)
    inertias.append(kmeans.inertia_)  # Сумма расстояний до ближайших центроид
    silhouettes.append(silhouette_score(X, kmeans.labels_))

# Визуализация результатов
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(range_n_clusters, inertias, marker='o')
plt.title('Метод локтя (Inertia)')
plt.xlabel('Число кластеров')
plt.ylabel('Суммарная внутрикластерная сумма квадратов')

plt.subplot(1, 2, 2)
plt.plot(range_n_clusters, silhouettes, marker='o')
plt.title('Silhouette Analysis')
plt.xlabel('Число кластеров')
plt.ylabel('Средний Silhouette Coefficient')
plt.show()

# 3. Выбор оптимального числа кластеров (оптимально около 3)
optimum_k = 3
kmeans_opt = KMeans(n_clusters=optimum_k, init='k-means++', random_state=42)
kmeans_opt.fit(X)

# 4. Визуализация кластеров
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=kmeans_opt.labels_, cmap='viridis', s=50)
plt.scatter(kmeans_opt.cluster_centers_[:, 0], kmeans_opt.cluster_centers_[:, 1],
            marker='x', s=200, linewidths=3, color='red', label='Центры кластеров')
plt.title('Результат кластеризации K-means')
plt.legend()
plt.show()

# 5. Оценка качества кластеризации
silhouette_avg = silhouette_score(X, kmeans_opt.labels_)
calinski_harabasz = calinski_harabasz_score(X, kmeans_opt.labels_)
print(f"Average Silhouette Score: {silhouette_avg:.4f}")
print(f"Calinski-Harabasz Index: {calinski_harabasz:.4f}")
```

9. Проверка гипотез:
   - Тест на однородность (Homogeneity test)
   - Тест на равенство средних (Mean equality test)

```python
# Тест на однородность
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency

# 1. Генерация синтетических данных (таблица сопряжённости)
np.random.seed(42)
categories = ["Category1", "Category2"]
features = ["Feature1", "Feature2"]
table_shape = (len(categories), len(features))

observed_freq = np.random.randint(10, 100, table_shape)
contingency_table = pd.DataFrame(observed_freq, columns=features, index=categories)

# 2. Проведение теста Хи-квадрат
chi2_stat, p_value, dof, expected_freq = chi2_contingency(contingency_table)

# 3. Визуализация данных
plt.figure(figsize=(10, 6))
sns.heatmap(expected_freq, annot=True, fmt=".2f", cmap="Blues", cbar=False)
plt.title("Ожидаемые частоты (Хи-квадрат)")
plt.show()

# 4. Вывод результатов
alpha = 0.05  # Уровень значимости
if p_value < alpha:
    conclusion = "Имеются доказательства зависимости переменных."
else:
    conclusion = "Независимость переменных не опровергнута."

report = f"""
Таблица сопряжённости:
{contingency_table.to_string()}

Статистика Хи-квадрат: {chi2_stat:.4f}
P-значение: {p_value:.4f}
Степени свободы: {dof}
Заключение: {conclusion}
"""
print(report)
```

10. Методы машинного обучения:
    - Случайный лес (Random forest)
    - Деревья решений (Decision trees)

```python
# Случайный лес
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
from sklearn.tree import export_graphviz
import graphviz

# 1. Генерация синтетических данных
np.random.seed(42)
X, y = make_classification(n_samples=1000, n_features=20, n_informative=5, n_redundant=5, class_sep=0.8, flip_y=0.1)

# Балансировка классов (если необходимо)
unique_classes, counts = np.unique(y, return_counts=True)
print(f"Classes Distribution: {dict(zip(unique_classes, counts))}")

# 2. Разделение на тренировочные и тестовые данные
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# 3. Настройка параметров с помощью GridSearchCV
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True, False]
}

rf = RandomForestClassifier(random_state=42)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
grid_search = GridSearchCV(rf, param_grid, cv=cv, scoring='accuracy', verbose=1, n_jobs=-1)
grid_search.fit(X_train, y_train)

# Лучшие параметры
best_params = grid_search.best_params_
print(f"Best Parameters: {best_params}")

# 4. Обучение модели с лучшими параметрами
best_rf = RandomForestClassifier(**best_params, random_state=42)
best_rf.fit(X_train, y_train)

# 5. Прогнозирование и оценка точности
y_pred = best_rf.predict(X_test)
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy on test set: {acc:.4f}")

# Матрица путаницы
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(cm, display_labels=["Class 0", "Class 1"])
disp.plot()
plt.show()

# 6. Визуализация важностей признаков
feature_importances = best_rf.feature_importances_
sorted_idx = feature_importances.argsort()[::-1]
plt.barh(range(len(sorted_idx)), feature_importances[sorted_idx], align='center')
plt.yticks(range(len(sorted_idx)), [f"Feature_{i}" for i in sorted_idx])
plt.xlabel("Importance")
plt.title("Важность признаков")
plt.show()

# 7. Визуализация одного дерева (первое дерево из леса)
dot_data = export_graphviz(best_rf.estimators_[0], out_file=None, filled=True, rounded=True, special_characters=True)
graph = graphviz.Source(dot_data)
graph.render(view=True)
```
